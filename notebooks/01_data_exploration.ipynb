```python
# Data manipulation
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# File management
from pathlib import Path

# Display settings
pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.float_format", "{:.2f}".format)

# Visualization style
sns.set_theme(style="whitegrid")

print("Libraries imported successfully!")
```

```python
# Locate the raw data folder
data_path = Path("../data/raw/house_rentals.csv")

# Check whether the file exists
if not data_path.exists():
    raise FileNotFoundError(
        f"Dataset not found: {data_path.resolve()}"
    )

# Load CSV
df = pd.read_csv(data_path)

print("Dataset loaded successfully!")
```

```python
rows, columns = df.shape

print(f"Number of records: {rows:,}")
print(f"Number of columns: {columns}")
```

```python
print("Dataset columns:\n")

for column in df.columns:
    print(column)
```

```python
missing_values = df.isnull().sum()

missing_summary = pd.DataFrame({
    "Missing Values": missing_values,
    "Missing Percentage": (
        missing_values / len(df) * 100
    )
})

missing_summary = missing_summary.sort_values(
    by="Missing Percentage",
    ascending=False
)

display(missing_summary)
```

```python
plt.figure(figsize=(12, 6))

sns.heatmap(
    df.isnull(),
    cbar=False,
    yticklabels=False,
    cmap="viridis"
)

plt.title("Missing Values in Ghana Rental Dataset")
plt.xlabel("Dataset Columns")
plt.show()
```

```python
duplicate_count = df.duplicated().sum()

print(f"Duplicate records: {duplicate_count}")
```

```python
df.describe()
```

```python
if "price" in df.columns:

    print("Rental price statistics:")

    print(df["price"].describe())

else:
    print("Price column not found. Check the column names.")
```

```python
if "price" in df.columns:

    plt.figure(figsize=(10, 5))

    sns.histplot(
        df["price"].dropna(),
        bins=40,
        kde=True
    )

    plt.title("Distribution of Rental Prices in Ghana")
    plt.xlabel("Rental Price (GHS)")
    plt.ylabel("Number of Listings")

    plt.show()
```

```python
regional_prices = (
    df.groupby("region")["price"]
    .agg(["count", "mean", "median"])
    .sort_values("median", ascending=False)
)

display(regional_prices)
```

```python
# Sorting Out House Type Category
if "category" in df.columns:

    print(df["category"].value_counts())

    plt.figure(figsize=(10, 5))

    sns.countplot(
        data=df,
        y="category",
        order=df["category"].value_counts().index
    )

    plt.title("Distribution of Property Categories")
    plt.xlabel("Number of Listings")
    plt.ylabel("Property Category")

    plt.show()
```

```python
for column in ["bedrooms", "bathrooms"]:

    if column in df.columns:

        print(f"\n{column.upper()} DISTRIBUTION")

        print(df[column].value_counts().sort_index())
```

```python
# Create a summary of the dataset
summary = {
    "Total Listings": len(df),
    "Total Columns": len(df.columns),
    "Duplicate Records": df.duplicated().sum(),
    "Total Missing Values": df.isnull().sum().sum()
}

summary_df = pd.DataFrame(
    summary.items(),
    columns=["Metric", "Value"]
)

display(summary_df)
```

```python
print("Listing type distribution:")

display(df["listing_type"].value_counts())
```

```python
print("Floor area statistics:")

display(df["floor_area"].describe())

print("\nLargest floor areas:")

display(
    df.nlargest(15, "floor_area")[
        ["name", "price", "bedrooms", "floor_area", "locality"]
    ]
)
```

```python
print("Most expensive properties:")

display(
    df.nlargest(15, "price")[
        ["name", "price", "category", "bedrooms",
         "bathrooms", "floor_area", "locality"]
    ]
)
```

```python
print("Cheapest properties:")

display(
    df.nsmallest(15, "price")[
        ["name", "price", "category", "bedrooms",
         "bathrooms", "floor_area", "locality"]
    ]
)
```
