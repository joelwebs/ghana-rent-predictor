```python
import pandas as pd
import numpy as np
from pathlib import Path

print("Libraries imported successfully!")
```

```python
DATA_PATH = Path("../data/raw/house_rentals.csv")

df = pd.read_csv(DATA_PATH)

print("Dataset loaded successfully!")
print("Shape:", df.shape)

df.head()
```

```python
clean_df = df.copy()

print("Working copy created.")
```

```python
missing_values = clean_df.isnull().sum()

print("Missing values per column:")
print(missing_values)

print("\nTotal missing values:", clean_df.isnull().sum().sum())
```

```python
duplicate_count = clean_df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
duplicate_urls = clean_df["url"].duplicated().sum()

print("Duplicate listing URLs:", duplicate_urls)
```

```python
clean_df[clean_df["url"].duplicated(keep=False)].sort_values("url").head(20)
```

```python
numerical_columns = [
    "price",
    "bedrooms",
    "bathrooms",
    "floor_area"
]

for column in numerical_columns:
    print(f"\n{column}")
    print("Minimum:", clean_df[column].min())
    print("Maximum:", clean_df[column].max())
    print("Unique values:", clean_df[column].nunique())
```

```python
print("Largest floor areas:")
display(clean_df.nlargest(15, "floor_area")[
    ["name", "price", "bedrooms", "bathrooms", "floor_area", "location"]
])
```

```python
print("Highest rental prices:")
display(clean_df.nlargest(15, "price")[
    ["name", "price", "bedrooms", "bathrooms", "floor_area", "location"]
])
```

```python
print("Highest rental prices:")
display(clean_df.nsmallest(15, "price")[
    ["name", "price", "bedrooms", "bathrooms", "floor_area", "location"]
])
```

```python
categorical_columns = [
    "listing_type",
    "category",
    "condition",
    "region",
    "locality",
    "is_furnished"
]

for column in categorical_columns:
    print(f"\n{column}:")
    print(clean_df[column].value_counts(dropna=False).head(20))
```

```python
print("Prices less than or equal to zero:",
      (clean_df["price"] <= 0).sum())

print("Bedrooms less than or equal to zero:",
      (clean_df["bedrooms"] <= 0).sum())

print("Bathrooms less than or equal to zero:",
      (clean_df["bathrooms"] <= 0).sum())

print("Floor areas less than or equal to zero:",
      (clean_df["floor_area"] <= 0).sum())
```

```python
print("Latitude range:")
print(clean_df["lat"].min(), "to", clean_df["lat"].max())

print("\nLongitude range:")
print(clean_df["lng"].min(), "to", clean_df["lng"].max())
```

## Saving the cloned and cleaned Data from Original Dataset ##

```python
OUTPUT_PATH = Path("../data/processed/rentals_investigation.csv")

OUTPUT_PATH.parent.mkdir(parents=True, exist_ok=True)

clean_df.to_csv(OUTPUT_PATH, index=False)

print("Investigation dataset saved successfully!")
```
